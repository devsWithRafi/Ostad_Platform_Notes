    Write a Python program that takes two sets from the user and computes their union, intersection, and difference.

    A = {1, 2, 3, 4, 5}
    B = {4, 5, 6, 7, 8}

```python
A_input = input("Enter Your (A) Set: ").split()
B_input = input("Enter Your (B) Set: ").split()

A = [int(i) for i in A_input]
B = [int(i) for i in B_input]

A,B = set(A), set(B)
```

```python
Union = A | B
print(Union)
```

```python
Intersection = A & B
print(Intersection)
```

```python
Difference = A - B
print(Difference)
```
