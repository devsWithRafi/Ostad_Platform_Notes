### **Problem: 01**

    Write a function that takes two numbers as input and
    returns their average. Call the function with different values

```python
# Answer:
def average_value(num1, num2):
  average = (num1 + num2) / 2
  return average

print(average_value(10, 20))
print(average_value(5, 15))
print(average_value(7, 13))
print(average_value(100, 200))
```

### **Problem: 02**

    Write a function that sorts a list of strings alphabetically.
    ["apple", "banana", "cherry", "kiwi", "grape"]

```python
# Answer:
def sorts_a_list():
  unsorted_lists = ["apple", "banana", "cherry", "kiwi", "grape"]
  sorted_lists = sorted(unsorted_lists, reverse=False)
  return sorted_lists

print(sorts_a_list())
```

### **Problem: 03**

    Write a Python program that creates a new list containing only
    the duplicate elements from the given list: [1, 5, 6, 5, 1, 2, 3].

```python
# Answer:
lists =  [1, 5, 6, 5, 1, 2, 3]

duplicates = []

for i in range(len(lists)):
  for j in range(i+1, len(lists)):
    if lists[i] == lists[j]:
      duplicates.append(lists[j])

print(duplicates)
```
