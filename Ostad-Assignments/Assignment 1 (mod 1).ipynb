**Solution of Problem 01**

```python
def check_even_or_odd():

    get_user_input = int(input("Enter a number: "))
    if get_user_input % 2 == 0:
        print(f"'{get_user_input}' is a Even Number")
    else:
        print(f"'{get_user_input}' is a Odd Number")

check_even_or_odd()
```

**Solution of Problem 02**

```python
def calculation():

    first_num = int(input("Enter your 1st number: "))
    operator = input("Enter operator (+, -, *, /) number: ")
    second_num = int(input("Enter your 2nd number: "))

    if operator == '+': result = first_num + second_num
    elif operator == '-': result = first_num - second_num
    elif operator == '*': result = first_num * second_num
    elif operator == '/':
        if second_num == 0: return "Can't divide by Zero!"
        else: result = first_num / second_num
    else: return "Command not found!"

    return f"Result = {result}"

calculation()
```
