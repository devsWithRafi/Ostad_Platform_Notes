```python
# Creating subplots with plt.subplot()
x2 = np.linspace(0, 10, 100)
```

```python


plt.figure(figsize=(15, 10))

# First subplot
plt.subplot(2, 2, 1)  # 2 rows, 2 columns, position 1
plt.plot(x2, np.sin(x2), 'b-', linewidth=2)
plt.title('Sine Wave')
plt.grid(True, alpha=0.3)


# Second subplot
plt.subplot(2, 2, 2)  # position 2
plt.title("Student's Study Trend")
plt.xlabel("nth Day")
plt.ylabel("No. of Hours Studied")
plt.grid(True);
plt.scatter(x,y)



# Third subplot
plt.subplot(2, 2, 3)  # position 3
plt.plot(x2, np.tan(x2), 'g-', linewidth=2)
plt.title('Tangent Wave')
plt.ylim(-5, 5)
plt.grid(True, alpha=0.3)


# Fourth subplot
plt.subplot(2, 2, 4)  # position 4
plt.plot(x2, np.exp(-x2/5) * np.sin(x2), 'm-', linewidth=2)
plt.title('Damped Sine Wave')
plt.grid(True, alpha=0.3)

plt.savefig("plot.png")
```

```python
# Exploring colors, markers, and line styles
x = np.linspace(0, 10, 20)
y1 = x
y2 = x**2
y3 = x**0.5

plt.figure(figsize=(12, 8))

# Different colors and line styles
plt.plot(x, y1, color='red', linestyle='-', marker='o', markersize=6, label='Linear (y=x)')
plt.plot(x, y2, color='blue', linestyle='--', marker='s', markersize=6, label='Quadratic (y=x²)')
plt.plot(x, y3, color='green', linestyle=':', marker='^', markersize=8, label='Square root (y=√x)')

# Customizing the plot
plt.title('Different Line Styles and Markers', fontsize=16, pad=20)
plt.xlabel('X values', fontsize=12)
plt.ylabel('Y values', fontsize=12)
plt.legend(loc='upper left', fontsize=10)
plt.grid(True, alpha=0.3, linestyle=':')

# Set axis limits
plt.xlim(0, 10)
plt.ylim(0, 100)

plt.show()

print("Colors can be specified as names ('red'), hex codes ('#FF0000'), or RGB tuples!")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Create a figure and a single Axes
fig, ax = plt.subplots()


# Plot some data on the Axes
ax.plot(np.random.rand(10))
ax.set_title('Single Plot')



# Create a figure with a 2x2 grid of Axes
fig, axs = plt.subplots(2, 2)

# Plot on individual Axes
axs[0, 0].plot(np.random.rand(10))
axs[0, 0].set_title('Top-Left')

axs[0, 1].plot(np.sin(np.linspace(0, 2*np.pi, 100)))
axs[0, 1].set_title('Top-Right')

axs[1, 0].plot(np.cos(np.linspace(0, 2*np.pi, 100)))
axs[1, 0].set_title('Bottom-Left')

axs[1, 1].scatter(np.random.rand(10), np.random.rand(10))
axs[1, 1].set_title('Bottom-Right')

fig.suptitle('Multiple Subplots Example') # Add a title to the entire figure
plt.tight_layout() # Adjust subplot parameters for a tight layout
plt.show()
```

```python
# Advanced subplots with plt.subplots()
fig, axes = plt.subplots(2, 3, figsize=(18, 10))

# Generate sample data
np.random.seed(42)
data1 = np.random.normal(0, 1, 1000)
data2 = np.random.exponential(2, 1000)
x = np.linspace(0, 10, 50)

# Plot 1: Line plot
axes[0, 0].plot(x, np.sin(x), 'b-', linewidth=2)
axes[0, 0].set_title('Sine Wave')
axes[0, 0].grid(True, alpha=0.3)

# Plot 2: Histogram
axes[0, 1].hist(data1, bins=30, alpha=0.7, color='skyblue')
axes[0, 1].set_title('Normal Distribution')
axes[0, 1].grid(True, alpha=0.3)

# Plot 3: Scatter plot
axes[0, 2].scatter(data1[:100], data2[:100], alpha=0.6, color='red')
axes[0, 2].set_title('Scatter Plot')
axes[0, 2].grid(True, alpha=0.3)

# Plot 4: Bar chart
categories = ['A', 'B', 'C', 'D']
values = [23, 45, 56, 78]
axes[1, 0].bar(categories, values, color=['red', 'green', 'blue', 'orange'])
axes[1, 0].set_title('Bar Chart')

# Plot 5: Box plot
axes[1, 1].boxplot([data1, data2], labels=['Normal', 'Exponential'])
axes[1, 1].set_title('Box Plot Comparison')

# Plot 6: Pie chart
sizes = [30, 25, 25, 20]
axes[1, 2].pie(sizes, labels=categories, autopct='%1.1f%%')
axes[1, 2].set_title('Pie Chart')

plt.tight_layout()
plt.show()

print("plt.subplots() returns figure and axes objects for more control!")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# data from https://allisonhorst.github.io/palmerpenguins/

species = (
    "Adelie\n $\\mu=$3700.66g",
    "Chinstrap\n $\\mu=$3733.09g",
    "Gentoo\n $\\mu=5076.02g$",
)
weight_counts = {
    "Below": np.array([70, 31, 58]),
    "Above": np.array([82, 37, 66]),
}

width = .2

fig, ax = plt.subplots(figsize=(10, 5))
bottom = np.zeros(3)

for boolean, weight_count in weight_counts.items():
    p = ax.bar(species, weight_count, width, label=boolean, bottom=bottom)
    bottom += weight_count

ax.set_title("Number of penguins with above average body mass")
ax.legend(loc="upper right")

plt.show()
```

```python
# data from https://allisonhorst.github.io/palmerpenguins/

import matplotlib.pyplot as plt
import numpy as np

species = ("Adelie", "Chinstrap", "Gentoo")
penguin_means = {
    'Bill Depth': (18.35, 18.43, 14.98),
    'Bill Length': (38.79, 48.83, 47.50),
    'Flipper Length': (189.95, 195.82, 217.19),
}

x = np.arange(len(species))  # the label locations
print(x)
width = 0.25  # the width of the bars
multiplier = 0

fig, ax = plt.subplots(figsize=(10,5))

for attribute, measurement in penguin_means.items():
    offset = width * multiplier *2

    rects = ax.bar(x+offset, measurement, width, label=attribute)
    ax.bar_label(rects, padding=3)
    multiplier += 1

# Add some text for labels, title and custom x-axis tick labels, etc.
ax.set_ylabel('Length (mm)')
ax.set_title('Penguin attributes by species')
ax.set_xticks(x + width, species)
ax.legend(loc='upper left', ncols=3)
ax.set_ylim(0, 250)

plt.show()
```
